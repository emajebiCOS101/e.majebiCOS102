```python
print("Equation Root Calculator")
print("1. Quadratic Equation")
print("2. Cubic Equation")
print("3. Quartic Equation")

option = int(input("Enter your choice: "))

if option == 1:
    print("Quadratic Equation: Ax² + Bx + C = 0")

    A = float(input("Input A: "))
    B = float(input("Input B: "))
    C = float(input("Input C: "))

    discriminant = (B ** 2) - (4 * A * C)

    root1 = (-B + discriminant ** 0.5) / (2 * A)
    root2 = (-B - discriminant ** 0.5) / (2 * A)

    print("1st root =", root1)
    print("2nd root =", root2)

elif option == 2:
    print("Cubic Equation: Ax³ + Bx² + Cx + D = 0")

    A = float(input("Input A: "))
    B = float(input("Input B: "))
    C = float(input("Input C: "))
    D = float(input("Input D: "))

    print("You entered:")
    print(A, "x³ +", B, "x² +", C, "x +", D, "= 0")
    print("Cubic root calculation is not available.")

elif option == 3:
    print("Quartic Equation: Ax⁴ + Bx³ + Cx² + Dx + E = 0")

    A = float(input("Input A: "))
    B = float(input("Input B: "))
    C = float(input("Input C: "))
    D = float(input("Input D: "))
    E = float(input("Input E: "))

    print("You entered:")
    print(A, "x⁴ +", B, "x³ +", C, "x² +", D, "x +", E, "= 0")
    print("Quartic root calculation is not available.")

else:
    print("Invalid option")
```
