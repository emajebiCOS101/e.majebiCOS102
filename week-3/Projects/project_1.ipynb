```python
import tkinter as tk
from tkinter import messagebox
import pandas as pd

data = pd.read_csv("GIG-logistics(in).csv")


def verify_employee():

    name = name_entry.get()
    department = department_entry.get()

    employee = data[
        (data["Name"].str.lower() == name.lower()) &
        (data["Department"].str.lower() == department.lower())
    ]

    if not employee.empty:

        department_members = data[
            data["Department"].str.lower() == department.lower()
        ]["Name"].tolist()

        members = "\n".join(department_members)

        messagebox.showinfo(
            "Employee Found",
            f"Welcome {name}!\n\n"
            f"You are a member of the {department} department.\n\n"
            f"Other members in this department are:\n{members}"
        )

    else:
        messagebox.showerror(
            "Employee Not Found",
            "Sorry, employee does not exist."
        )


root = tk.Tk()
root.title("GIG Logistics Employee Verification System")
root.geometry("500x300")


heading = tk.Label(
    root,
    text="GIG Logistics Employee Verification",
    font=("Times New Roman", 14, "bold")
)
heading.pack(pady=10)


name_label = tk.Label(root, text="Enter Employee Name")
name_label.pack()

name_entry = tk.Entry(root, width=40)
name_entry.pack(pady=5)


department_label = tk.Label(root, text="Enter Department")
department_label.pack()

department_entry = tk.Entry(root, width=40)
department_entry.pack(pady=5)


verify_button = tk.Button(
    root,
    text="Verify Employee",
    command=verify_employee,
    bg="pink",
    fg="red"
)

verify_button.pack(pady=20)

root.mainloop()
```
